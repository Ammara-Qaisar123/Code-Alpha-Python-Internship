## 🎯 Objective

The objective of this project is to create a simple program that calculates the total value of a stock portfolio.

The program allows the user to:

* Select stocks from a predefined list.
* Enter the quantity of each stock.
* Calculate the value of each investment.
* Calculate the total portfolio value.
* Display a summary of the investments.


## Step 1: Create Stock Price Dictionary

A dictionary is used to store the names of stocks and their corresponding
prices.

The prices are predefined manually, as required by the project.

```python
stock_prices = {
    "AAPL": 180,
    "TSLA": 250,
    "MSFT": 420,
    "GOOGL": 170,
    "AMZN": 190
}

print("Available Stocks:")
for stock, price in stock_prices.items():
    print(stock, "=", "$", price)
```

## Step 2: Initialize Portfolio

An empty dictionary is created to store the stocks selected by the user
and their quantities.

```python
portfolio = {}
```

## Step 3: Get Number of Stocks

The user is asked how many different stocks they want to add to their
portfolio.

```python
number_of_stocks = int(input("How many different stocks do you want to add? "))
```

## Step 4: Enter Stock Names and Quantities

The user enters a stock symbol and the number of shares they own.

The program checks whether the entered stock is available in the predefined
stock-price dictionary.

```python
for i in range(number_of_stocks):

    stock = input("\nEnter stock symbol: ").upper()

    if stock in stock_prices:

        quantity = int(input("Enter quantity: "))

        portfolio[stock] = quantity

    else:
        print("Stock not available.")4
```

## Step 5: Calculate Investment Value

The value of each stock investment is calculated using:

Investment Value = Stock Price × Quantity

The value of all investments is then added together to calculate the
total portfolio value.

```python
total_investment = 0

print("\n===== PORTFOLIO SUMMARY =====")

for stock, quantity in portfolio.items():

    price = stock_prices[stock]

    investment_value = price * quantity

    total_investment += investment_value

    print(
        stock,
        "| Price: $", price,
        "| Quantity:", quantity,
        "| Value: $", investment_value
    )

print("\nTotal Investment Value: $", total_investment)
```

## 🧮 Example Calculation

Suppose the user enters:

- AAPL → 5 shares
- TSLA → 2 shares

The calculations will be:

AAPL = 180 × 5 = $900

TSLA = 250 × 2 = $500

Total Investment = $900 + $500 = $1,400

## Step 6: Save Portfolio Summary

The portfolio summary can optionally be saved to a text file.

This demonstrates basic Python file handling.

```python
with open("portfolio_summary.txt", "w") as file:

    file.write("===== STOCK PORTFOLIO SUMMARY =====\n\n")

    for stock, quantity in portfolio.items():

        price = stock_prices[stock]
        investment_value = price * quantity

        file.write(
            f"{stock} | Price: ${price} | "
            f"Quantity: {quantity} | "
            f"Value: ${investment_value}\n"
        )

    file.write(f"\nTotal Investment Value: ${total_investment}")

print("Portfolio summary saved successfully!")
```
