### Project Description

This project is a simple rule-based chatbot developed using Python.

The chatbot receives text input from the user and provides predefined
responses based on the user's message.

The chatbot can respond to greetings, questions about how it is doing,
and a goodbye message.

## Objective

The objective of this project is to develop a simple interactive chatbot
using basic Python programming concepts.

The chatbot should:
- Accept input from the user.
- Recognize predefined messages.
- Provide appropriate responses.
- Continue chatting until the user enters "bye".

## Step 1: Create a Chatbot Function

A function named `chatbot_response()` is created to process the user's
message and return an appropriate response.

```python
def chatbot_response(user_input):

    user_input = user_input.lower()

    if user_input == "hello":
        return "Hi!"

    elif user_input == "how are you":
        return "I'm fine, thanks!"

    elif user_input == "bye":
        return "Goodbye!"

    else:
        return "Sorry, I don't understand."
```

## Step 2: Test the Chatbot Function

We can test the function by providing different messages and observing
the responses.

```python
print(chatbot_response("hello"))
print(chatbot_response("how are you?"))
print(chatbot_response("bye"))
```

## Step 3: Create an Interactive Chat Loop

A `while` loop is used so that the chatbot can continue communicating
with the user.

The loop stops when the user enters "bye".

```python
print("===== BASIC CHATBOT =====")
print("Type 'bye' to exit.")

while True:

    user_input = input("You: ")

    response = chatbot_response(user_input)

    print("Bot:", response)

    if user_input.lower() == "bye":
        break
```
