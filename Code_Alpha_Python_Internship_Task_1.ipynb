## Hangman Game

### Project Description

This project is a simple text-based Hangman game developed using Python.
The player has to guess a hidden word one letter at a time.

The game selects a word randomly from a predefined list of five words.
The player is allowed a maximum of six incorrect guesses.

### Technologies Used
- Python
- Random module
- Lists
- Strings
- While loop
- If-else statements
- User input

## Objective

The objective of this project is to create a simple Hangman game that allows
the player to guess a randomly selected word.

The game should:
- Select a word randomly.
- Display blanks for unknown letters.
- Allow the player to guess one letter at a time.
- Track incorrect guesses.
- Allow a maximum of six incorrect guesses.
- Display a winning or game-over message.

## Step 1: Import Required Library

The `random` module is used to randomly select a word from our predefined
list of words.

```python
import random
```

## Step 2: Create a List of Words

A list containing five predefined words is created. The computer will
randomly select one word from this list for the player to guess.

```python
words = ["python", "computer", "programming", "science", "developer"]

print("Available words:", words)
```

## Step 3: Select a Random Word

The `random.choice()` function selects one word randomly from the predefined
list.

This selected word becomes the secret word that the player needs to guess.

```python
word = random.choice(words)

print("A secret word has been selected!")
```

## Step 4: Create Hidden Word

The selected word is hidden from the player.

An underscore `_` is created for every letter in the secret word.
The underscores will be replaced when the player correctly guesses letters.

```python
guessed_word = ["_"] * len(word)

print("Word:", " ".join(guessed_word))
```

## Step 5: Initialize Game Variables

We create variables to keep track of:

- Number of incorrect guesses
- Maximum allowed incorrect guesses
- Letters already guessed by the player

```python
incorrect_guesses = 0
max_guesses = 6
guessed_letters = []

print("Maximum incorrect guesses:", max_guesses)
```

## Step 6: Start the Game Loop

A `while` loop is used to keep the game running.

The game continues while:
1. The player has not reached six incorrect guesses.
2. There are still unknown letters in the word.

```python
while incorrect_guesses < max_guesses and "_" in guessed_word:

    print("\nWord:", " ".join(guessed_word))

    guess = input("Guess a letter: ").lower()

    if len(guess) != 1 or not guess.isalpha():
        print("Please enter only one letter.")
        continue

    if guess in guessed_letters:
        print("You already guessed that letter.")
        continue

    guessed_letters.append(guess)

    if guess in word:
        print("Correct guess!")

        for i in range(len(word)):
            if word[i] == guess:
                guessed_word[i] = guess

    else:
        incorrect_guesses += 1
        print("Wrong guess!")
        print("Incorrect guesses:", incorrect_guesses, "/", max_guesses)
```

## Step 7: Display Final Result

After the game loop finishes, the program checks whether the player
successfully guessed the complete word.

If there are no underscores remaining, the player wins.
Otherwise, the maximum number of incorrect guesses has been reached.

```python
if "_" not in guessed_word:
    print("\nCongratulations!")
    print("You guessed the word:", word)
else:
    print("\nGame Over!")
    print("The word was:", word)
```
